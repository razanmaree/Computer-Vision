```python
from google.colab import drive
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.ndimage import maximum_filter
from sklearn.cluster import DBSCAN
from sklearn.linear_model import LinearRegression

drive.mount('/content/drive/')
path = '/content/drive/My Drive/ex1/group_sketch/what_is_a_triangle.png'
input_img = cv2.imread(path)


# Edge map
gray = cv2.cvtColor(input_img, cv2.COLOR_BGR2GRAY)
edgesCanny = cv2.Canny(gray, 200, 300)
edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)

def hough_transform(edges):
    rows, cols = edges.shape
    max_dist = int(np.hypot(rows, cols))
    thetas = np.deg2rad(np.arange(-90, 89))
    rhos = np.linspace(-max_dist, max_dist, 2 * max_dist)
    accumulator = np.zeros((2 * max_dist, len(thetas)), dtype=np.int32)
    edge_points = np.argwhere(edges)
    point_votes = {}
    for y, x in edge_points:
        for theta_idx, theta in enumerate(thetas):
            rho = int(x * np.cos(theta) + y * np.sin(theta)) + max_dist
            accumulator[rho, theta_idx] += 1
            if (rho, theta_idx) not in point_votes:
                point_votes[(rho, theta_idx)] = []
            point_votes[(rho, theta_idx)].append((x, y))
    accumulator_image = np.uint8(accumulator)
    return accumulator_image, rhos, thetas, accumulator, point_votes

# Call the hough_transform function
hough_transform_result, rhos, thetas, accumulator, point_votes = hough_transform(edgesCanny)

# Normalize the accumulator
accumulator_normalized = (hough_transform_result / np.max(hough_transform_result)) * 255
accumulator_normalized = np.uint8(accumulator_normalized)
accumulator_normalized[accumulator_normalized != 0] += 70

def find_lines(accumulator, threshold, rhos, thetas):
    local_max = maximum_filter(accumulator, size=10) == accumulator
    accumulator_peaks = local_max & (accumulator > threshold)
    rho_idxs, theta_idxs = np.nonzero(accumulator_peaks)
    rhos_peaks = rhos[rho_idxs]
    thetas_peaks = thetas[theta_idxs]
    return rhos_peaks, thetas_peaks, rho_idxs, theta_idxs

# Define a threshold for peak detection
threshold = 0.20 * np.max(hough_transform_result)

# Find peaks in the Hough Transform space
rhos_peaks, thetas_peaks, rho_idxs, theta_idxs = find_lines(hough_transform_result, threshold, rhos, thetas)


rows, cols = edgesCanny.shape
def get_line_endpoints(rho, theta, cols, rows):
    """ Get the endpoints of a line given rho and theta """
    cos_t = np.cos(theta)
    sin_t = np.sin(theta)
    x0 = cos_t * rho
    y0 = sin_t * rho
    points = []
    if sin_t != 0:
        points.append((0, int(rho / sin_t)))
        points.append((cols - 1, int((rho - (cols - 1) * cos_t) / sin_t)))
    if cos_t != 0:
        points.append((int(rho / cos_t), 0))
        points.append((int((rho - (rows - 1) * sin_t) / cos_t), rows - 1))
    valid_points = [(x, y) for x, y in points if 0 <= x < cols and 0 <= y < rows]
    return valid_points

detected_lines = []
line_equations = []

for rho, theta in zip(rhos_peaks, thetas_peaks):
    points = get_line_endpoints(rho, theta, cols, rows)
    if len(points) == 2:
        detected_lines.append((points[0][0], points[0][1], points[1][0], points[1][1], rho, theta))
        line_equations.append((rho, theta))


# Function to calculate overlap with limited line length
def calculate_overlap(line_canvas, edge_map, line_length):
    overlap = np.zeros(edge_map.shape, dtype=np.uint8)
    nonzero_indices = cv2.findNonZero(line_canvas)

    if nonzero_indices is not None:
        line_mask = np.zeros_like(edge_map)
        cv2.polylines(line_mask, [nonzero_indices], isClosed=False, color=255, thickness=1)

        # Mask the line length
        masked_line = np.zeros_like(edge_map)
        x1, y1 = nonzero_indices[0][0]
        x2, y2 = nonzero_indices[-1][0]
        cv2.line(masked_line, (x1, y1), (x2, y2), 255, 1)

        # Calculate overlap
        overlap = cv2.bitwise_and(masked_line, edge_map)

    return np.count_nonzero(overlap)

# Grouping lines based on overlapping regions
def group_lines_by_overlap(lines, edge_map,line_equations, threshold=50):
    grouped_lines = []
    overlaps = []

    for line in lines:
        blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
        x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
        cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
        overlap = calculate_overlap(blank_canvas, edge_map, line_length=np.hypot(x2 - x1, y2 - y1))
        overlaps.append(overlap)

    # Sort lines by overlap
    sorted_lines = [line for _, line in sorted(zip(overlaps, lines), key=lambda pair: pair[0], reverse=True)]

    while sorted_lines:
        # Take the line with the maximum overlap
        best_line = sorted_lines.pop(0)
        best_blank = np.zeros(edge_map.shape, dtype=np.uint8)
        #x1, y1, x2, y2 = best_line
        x1, y1, x2, y2 = best_line[0], best_line[1], best_line[2], best_line[3]
        cv2.line(best_blank, (x1, y1), (x2, y2), 255, 1)

        # Group lines that overlap significantly with the best line
        group = [best_line]
        to_remove = []
        for line in sorted_lines:
            blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
            x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
            cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
            overlap = calculate_overlap(best_blank, blank_canvas, line_length=np.hypot(x2 - x1, y2 - y1))
            if overlap > threshold:
                group.append(line)
                to_remove.append(line)

        # Remove grouped lines from the list
        for line in to_remove:
            sorted_lines.remove(line)

        # Keep the best line of the group (we already have it as the best_line)
        grouped_lines.append(best_line)

    return grouped_lines


# Define a threshold for significant overlap
overlap_threshold = 10

# Group and filter lines based on overlap
grouped_lines = group_lines_by_overlap(detected_lines, edgesCanny ,line_equations, overlap_threshold)


unique_thetas =[]
unique_rhos = []
for line in grouped_lines:
    unique_rhos.append(line[4])
    unique_thetas.append(line[5])


def get_voted_points(point_votes, rho_idx, theta_idx):
    return point_votes.get((rho_idx, theta_idx), [])

def remove_outliers(points, eps=15, min_samples=5):
    if len(points) < min_samples:
        return points  # Not enough points to form a cluster, return all points
    clustering = DBSCAN(eps=eps, min_samples=min_samples).fit(points)
    labels = clustering.labels_
    # Keep points that are not classified as outliers (label != -1)
    filtered_points = [point for point, label in zip(points, labels) if label != -1]
    return filtered_points


from sklearn.linear_model import RANSACRegressor
def find_best_fitting_lines(unique_rhos, unique_thetas, point_votes):
    line_segments = []

    for i, (rho, theta) in enumerate(zip(unique_rhos, unique_thetas)):
        points = get_voted_points(point_votes, np.where(rhos == rho)[0][0], np.where(thetas == theta)[0][0])
        points = remove_outliers(points)
        if points:
            # Handle vertical lines separately
            if np.isclose(theta, 0) or np.isclose(theta, np.pi):
                x_min = min(p[0] for p in points)
                x_max = max(p[0] for p in points)
                line_segments.append(((x_min, points[0][1]), (x_max, points[-1][1]), (rho, theta)))
            else:
                # Fit a line to the points using RANSAC
                points = np.array(points)
                X = points[:, 0].reshape(-1, 1)
                y = points[:, 1]

                ransac = RANSACRegressor(LinearRegression(), min_samples=2, residual_threshold=5.0, max_trials=100)
                ransac.fit(X, y)
                inlier_mask = ransac.inlier_mask_

                if np.sum(inlier_mask) > 0:
                    x_inliers = X[inlier_mask]
                    y_inliers = y[inlier_mask]
                    x_start, x_end = x_inliers.min(), x_inliers.max()
                    y_start, y_end = ransac.predict([[x_start], [x_end]])
                    line_segments.append(((x_start, y_start), (x_end, y_end), (rho, theta)))
                else:
                    print(f'All points are outliers for line {i+1} with rho {rho} and theta {theta}')
        else:
            print(f'No points found for line {i+1} with rho {rho} and theta {theta}')
    return line_segments


# Find the best fitting lines
line_segments = find_best_fitting_lines(unique_rhos, unique_thetas,point_votes)


def remove_long_lines(line_segments, max_length=800,min_lenght=0,out=1):
  to_remove = []
  for i, ((x_start, y_start), (x_end, y_end), (rho, theta)) in enumerate(line_segments):
    line_length = np.sqrt((x_end - x_start) ** 2 + (y_end - y_start) ** 2)
    if line_length > max_length or (out < line_length < (out+1)) :
      to_remove.append(i)

  line_segments = [line_segments[i] for i in range(len(line_segments)) if i not in to_remove]
  return line_segments

line_segments = remove_long_lines(line_segments)


def find_side_intersection(p1, p2, p3, p4, window_size=57):  #57
    def vector(pt1, pt2):
        return (pt2[0] - pt1[0], pt2[1] - pt1[1])

    def cross_product(vect1, vect2):
        return vect1[0] * vect2[1] - vect1[1] * vect2[0]

    def enlarge(point1, point2, win_size):
        delta_x = point2[0] - point1[0]
        delta_y = point2[1] - point1[1]
        length = (delta_x**2 + delta_y**2)**0.5
        unit_x = delta_x / length
        unit_y = delta_y / length
        new_point1 = (point1[0] - unit_x * win_size, point1[1] - unit_y * win_size)
        new_point2 = (point2[0] + unit_x * win_size, point2[1] + unit_y * win_size)
        return (int(new_point1[0]), int(new_point1[1])), (int(new_point2[0]), int(new_point2[1]))

    vec_r = vector(enlarge(p1, p2, window_size)[0], enlarge(p1, p2, window_size)[1])
    vec_s = vector(enlarge(p3, p4, window_size)[0], enlarge(p3, p4, window_size)[1])
    vec_p3p1 = vector(enlarge(p1, p2, window_size)[0], enlarge(p3, p4, window_size)[0])
    cross_rs = cross_product(vec_r, vec_s)
    cross_p3p1r = cross_product(vec_p3p1, vec_r)

    if cross_rs == 0:
        return False, None  # Collinear or parallel

    t = cross_product(vec_p3p1, vec_s) / cross_rs
    u = cross_p3p1r / cross_rs

    if 0 <= t <= 1 and 0 <= u <= 1:
        intersection_point = (int(enlarge(p1, p2, window_size)[0][0] + t * vec_r[0]), int(enlarge(p1, p2, window_size)[0][1] + t * vec_r[1]))
        return True, intersection_point

    return False, None


def compute_distance(point1, point2):
    delta_x = point2[0] - point1[0]
    delta_y = point2[1] - point1[1]
    distance = (delta_x ** 2 + delta_y ** 2) ** 0.5
    return distance


def find_triangles(all_segments):

    triangles = []
    num_segments = len(all_segments)

    for i in range(num_segments - 2):
        for j in range(i + 1, num_segments - 1):
            for k in range(j + 1, num_segments):
                segment1 = all_segments[i]
                segment2 = all_segments[j]
                segment3 = all_segments[k]

                # Check if the three segments form a triangle
                intersect1, point1 = find_side_intersection(segment1[0], segment1[1], segment2[0], segment2[1])
                intersect2, point2 = find_side_intersection(segment2[0], segment2[1], segment3[0], segment3[1])
                intersect3, point3 = find_side_intersection(segment1[0], segment1[1], segment3[0], segment3[1])

                if intersect1 and intersect2 and intersect3:
                    flag = 0
                    if compute_distance(point1, point2) + compute_distance(point2, point3) <= compute_distance(point1, point3):
                        flag = 1

                    if compute_distance(point1, point2) + compute_distance(point1, point3) <= compute_distance(point2, point3):
                        flag = 1

                    if compute_distance(point2, point3) + compute_distance(point1, point3) <= compute_distance(point1, point2):
                        flag = 1

                    if flag == 0:
                        triangles.append((point1, point2, point3,segment1 ,segment2, segment3))

    return triangles

triangles = find_triangles(line_segments)


def filter_triangles(triangles,min_area=3300,max_area=8500, min_len=60, max_len=600,flag_is_mutual=True,flag_len=False):
    valid_triangles = []
    for triangle in triangles:
        flag=0
        p1, p2, p3 ,_,__,___= triangle
        x1, y1 = p1
        x2, y2 = p2
        x3, y3 = p3

        #check if the intersection point out of the image
        if x1 < 0 or x1 >= edgesCanny.shape[1] or x2 < 0 or x2 >= edgesCanny.shape[1] or x3 < 0 or x3 >= edgesCanny.shape[1]:
          flag=1
        if y1 < 0 or y1 >= edgesCanny.shape[0] or y2 < 0 or y2 >= edgesCanny.shape[0] or y3 < 0 or y3 >= edgesCanny.shape[0]:
          flag=1

        # Calculate the area of the triangle
        area = 0.5 * abs(x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

        # Check if the area is within the specified range
        if (min_area <= area <= max_area) and flag==0 :

            #check the lenght of the sides os each triangle
            side1_length = np.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
            side2_length = np.sqrt((x2 - x3) ** 2 + (y2 - y3) ** 2)
            side3_length = np.sqrt((x3 - x1) ** 2 + (y3 - y1) ** 2)
            if all(min_len <= s <= max_len for s in [side1_length, side2_length, side3_length]):
              valid_triangles.append(triangle)


    if flag_is_mutual:
      num_triangles = len(valid_triangles)
      to_remove = []
      for i in range(num_triangles):
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          for j in range(i + 1, num_triangles):
            p1, p2, p3 ,_,__,___= valid_triangles[i]
            p4, p5, p6 ,_,__,___= valid_triangles[j]
            thresh=10
            if (abs(p1[0]-p4[0])<thresh and abs(p1[1]-p4[1])<thresh) or (abs(p1[0]-p5[0])<thresh and abs(p1[1]-p5[1])<thresh) or (abs(p1[0]-p6[0])<thresh and abs(p1[1]-p6[1])<thresh) or\
            (abs(p2[0]-p4[0])<thresh and abs(p2[1]-p4[1])<thresh) or (abs(p2[0]-p5[0])<thresh and abs(p2[1]-p5[1])<thresh) or (abs(p2[0]-p6[0])<thresh and abs(p2[1]-p6[1])<thresh):
                area_j = 0.5 * abs(p4[0] * (p5[1] - p6[1]) + p5[0] * (p6[1] - p4[1]) + p6[0] * (p4[1] - p5[1]))
                if area_i > area_j:
                    to_remove.append(valid_triangles[i])

      valid_triangles = [t for t in valid_triangles if t not in to_remove]

    if flag_len:
      to_remove = []
      num_triangles = len(valid_triangles)
      #to_remove = []
      for i in range(num_triangles):
          p1, p2, p3 ,_,__,___= valid_triangles[i]
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          if 8480 < area_i < 14280  or 5000 < area_i <6000 or 7740 < area_i < 17770:
              to_remove.append(valid_triangles[i])
      valid_triangles = [t for t in valid_triangles if t not in to_remove]


    return valid_triangles

triangles = filter_triangles(triangles)


# Function to classify triangles
def classify_triangle(triangle, tol=15):
    def distance(p1, p2):
        return np.sqrt((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)

    a = distance(triangle[0], triangle[1])
    b = distance(triangle[1], triangle[2])
    c = distance(triangle[2], triangle[0])
    sides = sorted([a, b, c])

    # Check for right triangle
    if np.isclose(sides[0]**2 + sides[1]**2, sides[2]**2, atol=200):#200
        return 'right'

    # Check for equilateral
    if np.isclose(sides[0], sides[1], atol=10) and np.isclose(sides[1], sides[2], atol=10):#10 #10
        return 'equilateral'

    # Check for isosceles
    if np.isclose(sides[0], sides[1], atol=tol) or np.isclose(sides[1], sides[2], atol=tol) or np.isclose(sides[0], sides[2], atol=tol):
        return 'isosceles'

    # None of the above
    return 'other'

#------
def remove_close_triangles(triangles, threshold=2):
    # Initialize a list to store unique triangles
    unique_triangles = []

    # Iterate over each triangle in the list
    for triangle in triangles:
        # p1, p2, p3 = triangle
        p1, p2, p3 ,_,__,___= triangle

        is_unique = True

        # Compare this triangle with triangles already added to unique_triangles
        for unique_triangle in unique_triangles:
            q1, q2, q3 ,_,__,___= unique_triangle

            # Calculate distances between corresponding points of triangles
            dist1 = compute_distance(p1, q1)
            dist2 = compute_distance(p2, q2)
            dist3 = compute_distance(p3, q3)

            # Check if all distances are greater than the threshold
            if dist1 <= threshold and dist2 <= threshold and dist3 <= threshold:
                is_unique = False
                break

        # If the triangle is unique (distances are all greater than threshold), add it to unique_triangles
        if is_unique:
            unique_triangles.append(triangle)

    return unique_triangles

triangles = remove_close_triangles(triangles)
#------

def plot_colored_triangles(edgesCanny, triangles):
    plt.figure(figsize=(11, 11), dpi=100)
    plt.axis('off')
    plt.title('Detected triangles (color-coded)')

    # Create a copy of edgesCanny to avoid modifying the original
    edgesCanny_color = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2BGR)

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")
        if triangle_type == 'equilateral':
            color = (255, 0, 0)  # Blue in BGR
        elif triangle_type == 'isosceles':
            color = (0, 255, 0)  # Green in BGR
        elif triangle_type == 'right':
            color = (0, 0, 255)  # Red in BGR
        else:
            color = (255, 255, 255)  # White in BGR

        # Extract points from the triangle
        point1, point2, point3, _, _, _ = triangle

        # Draw the triangle using its points
        cv2.line(edgesCanny_color, point1, point2, color, 2)
        cv2.line(edgesCanny_color, point2, point3, color, 2)
        cv2.line(edgesCanny_color, point3, point1, color, 2)

    # Convert BGR image to RGB for displaying with matplotlib
    edgesCanny_color_rgb = cv2.cvtColor(edgesCanny_color, cv2.COLOR_BGR2RGB)

    plt.imshow(edgesCanny_color_rgb)
    plt.show()



def plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny):

    plt.figure(figsize=(5, 7), dpi=100)
    plt.imshow(accumulator_normalized, aspect='auto', cmap='gray')

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")
        if triangle_type == 'equilateral':
            color = 'blue'  # Blue
        elif triangle_type == 'isosceles':
            color = 'green'  # Green
        elif triangle_type == 'right':
            color = 'red'  # Red
        else:
            color = 'white'  # White

        # Extract rho, theta from the triangle sides
        _, _, _, side1, side2, side3 = triangle

        p1, p2, (rho1, theta1) = side1
        p3, p4, (rho2, theta2) = side2
        p5, p6, (rho3, theta3) = side3


        theta1 = np.rad2deg(theta1) + 90
        theta2 = np.rad2deg(theta2) + 90
        theta3 = np.rad2deg(theta3) + 90

        rows, cols = edgesCanny.shape
        max_dist = int(np.hypot(rows, cols))

        rho1 = rho1 + max_dist
        rho2 = rho2 + max_dist
        rho3 = rho3 + max_dist

        # Plot the peaks with the corresponding color
        plt.scatter([theta1, theta2, theta3], [rho1, rho2, rho3], c=color, s=20)

    plt.axis('off')
    plt.title('Hough Transform with color-coded triangle sides')
    plt.show()





#--------------------------> PLOTTING FINAL RESULTS <----------------------------------
# Display the input image
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(cv2.cvtColor(input_img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB
plt.axis('off')  # Hide the axes
plt.title('Input Image')
plt.show()

# Display the edge map
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(edgesCanny, cmap='gray')  # Use the gray colormap
plt.axis('off')  # Hide the axes
plt.title('Edge Map')
plt.show()

#Display the Hough transform
plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny)

#Display the detected lines
filtered_edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)
for line in grouped_lines:
    x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
    cv2.line(filtered_edges_canny_rgb, (x1, y1), (x2, y2), (255, 255, 255), 1)
plt.figure(figsize=(11, 11), dpi=100)
plt.axis('off')
plt.title('Detected Lines')
plt.imshow(filtered_edges_canny_rgb)
plt.show()


#Display colored triangles
plot_colored_triangles(edgesCanny, triangles)

```

```python
from google.colab import drive
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.ndimage import maximum_filter
from sklearn.cluster import DBSCAN
from sklearn.linear_model import LinearRegression

drive.mount('/content/drive/')
path = '/content/drive/My Drive/ex1/group_natural/il_570xN.3195615696_3srw.png'
input_img = cv2.imread(path)


# Edge map
gray = cv2.cvtColor(input_img, cv2.COLOR_BGR2GRAY)
edgesCanny = cv2.Canny(gray, 50, 150)
edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)

def hough_transform(edges):
    rows, cols = edges.shape
    max_dist = int(np.hypot(rows, cols))
    thetas = np.deg2rad(np.arange(-90, 89))
    rhos = np.linspace(-max_dist, max_dist, 2 * max_dist)
    accumulator = np.zeros((2 * max_dist, len(thetas)), dtype=np.int32)
    edge_points = np.argwhere(edges)
    point_votes = {}
    for y, x in edge_points:
        for theta_idx, theta in enumerate(thetas):
            rho = int(x * np.cos(theta) + y * np.sin(theta)) + max_dist
            accumulator[rho, theta_idx] += 1
            if (rho, theta_idx) not in point_votes:
                point_votes[(rho, theta_idx)] = []
            point_votes[(rho, theta_idx)].append((x, y))
    accumulator_image = np.uint8(accumulator)
    return accumulator_image, rhos, thetas, accumulator, point_votes

# Call the hough_transform function
hough_transform_result, rhos, thetas, accumulator, point_votes = hough_transform(edgesCanny)

# Normalize the accumulator
accumulator_normalized = (hough_transform_result / np.max(hough_transform_result)) * 255
accumulator_normalized = np.uint8(accumulator_normalized)
accumulator_normalized[accumulator_normalized != 0] += 70

def find_lines(accumulator, threshold, rhos, thetas):
    local_max = maximum_filter(accumulator, size=23) == accumulator
    accumulator_peaks = local_max & (accumulator > threshold)
    rho_idxs, theta_idxs = np.nonzero(accumulator_peaks)
    rhos_peaks = rhos[rho_idxs]
    thetas_peaks = thetas[theta_idxs]
    return rhos_peaks, thetas_peaks, rho_idxs, theta_idxs

# Define a threshold for peak detection
threshold = 0.35 * np.max(hough_transform_result)

# Find peaks in the Hough Transform space
rhos_peaks, thetas_peaks, rho_idxs, theta_idxs = find_lines(hough_transform_result, threshold, rhos, thetas)


rows, cols = edgesCanny.shape
def get_line_endpoints(rho, theta, cols, rows):
    """ Get the endpoints of a line given rho and theta """
    cos_t = np.cos(theta)
    sin_t = np.sin(theta)
    x0 = cos_t * rho
    y0 = sin_t * rho
    points = []
    if sin_t != 0:
        points.append((0, int(rho / sin_t)))
        points.append((cols - 1, int((rho - (cols - 1) * cos_t) / sin_t)))
    if cos_t != 0:
        points.append((int(rho / cos_t), 0))
        points.append((int((rho - (rows - 1) * sin_t) / cos_t), rows - 1))
    valid_points = [(x, y) for x, y in points if 0 <= x < cols and 0 <= y < rows]
    return valid_points

detected_lines = []
line_equations = []

for rho, theta in zip(rhos_peaks, thetas_peaks):
    points = get_line_endpoints(rho, theta, cols, rows)
    if len(points) == 2:
        detected_lines.append((points[0][0], points[0][1], points[1][0], points[1][1], rho, theta))
        line_equations.append((rho, theta))


# Function to calculate overlap with limited line length
def calculate_overlap(line_canvas, edge_map, line_length):
    overlap = np.zeros(edge_map.shape, dtype=np.uint8)
    nonzero_indices = cv2.findNonZero(line_canvas)

    if nonzero_indices is not None:
        line_mask = np.zeros_like(edge_map)
        cv2.polylines(line_mask, [nonzero_indices], isClosed=False, color=255, thickness=1)

        # Mask the line length
        masked_line = np.zeros_like(edge_map)
        x1, y1 = nonzero_indices[0][0]
        x2, y2 = nonzero_indices[-1][0]
        cv2.line(masked_line, (x1, y1), (x2, y2), 255, 1)

        # Calculate overlap
        overlap = cv2.bitwise_and(masked_line, edge_map)

    return np.count_nonzero(overlap)

# Grouping lines based on overlapping regions
def group_lines_by_overlap(lines, edge_map,line_equations, threshold=50):
    grouped_lines = []
    overlaps = []

    for line in lines:
        blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
        x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
        cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
        overlap = calculate_overlap(blank_canvas, edge_map, line_length=np.hypot(x2 - x1, y2 - y1))
        overlaps.append(overlap)

    # Sort lines by overlap
    sorted_lines = [line for _, line in sorted(zip(overlaps, lines), key=lambda pair: pair[0], reverse=True)]

    while sorted_lines:
        # Take the line with the maximum overlap
        best_line = sorted_lines.pop(0)
        best_blank = np.zeros(edge_map.shape, dtype=np.uint8)
        #x1, y1, x2, y2 = best_line
        x1, y1, x2, y2 = best_line[0], best_line[1], best_line[2], best_line[3]
        cv2.line(best_blank, (x1, y1), (x2, y2), 255, 1)

        # Group lines that overlap significantly with the best line
        group = [best_line]
        to_remove = []
        for line in sorted_lines:
            blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
            #x1, y1, x2, y2 = line
            x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
            cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
            overlap = calculate_overlap(best_blank, blank_canvas, line_length=np.hypot(x2 - x1, y2 - y1))
            if overlap > threshold:
                group.append(line)
                to_remove.append(line)

        # Remove grouped lines from the list
        for line in to_remove:
            sorted_lines.remove(line)

        # Keep the best line of the group (we already have it as the best_line)
        grouped_lines.append(best_line)

    #print(f'Length of grouped_lines is {len(grouped_lines)}')
    return grouped_lines


# Define a threshold for significant overlap
overlap_threshold = 5

# Group and filter lines based on overlap
grouped_lines = group_lines_by_overlap(detected_lines, edgesCanny ,line_equations, overlap_threshold)


unique_thetas =[]
unique_rhos = []
for line in grouped_lines:
    unique_rhos.append(line[4])
    unique_thetas.append(line[5])


def get_voted_points(point_votes, rho_idx, theta_idx):
    return point_votes.get((rho_idx, theta_idx), [])

def remove_outliers(points, eps=15, min_samples=10):
    if len(points) < min_samples:
        return points  # Not enough points to form a cluster, return all points
    clustering = DBSCAN(eps=eps, min_samples=min_samples).fit(points)
    labels = clustering.labels_
    # Keep points that are not classified as outliers (label != -1)
    filtered_points = [point for point, label in zip(points, labels) if label != -1]
    return filtered_points


from sklearn.linear_model import RANSACRegressor
def find_best_fitting_lines(unique_rhos, unique_thetas, point_votes):
    line_segments = []

    for i, (rho, theta) in enumerate(zip(unique_rhos, unique_thetas)):
        points = get_voted_points(point_votes, np.where(rhos == rho)[0][0], np.where(thetas == theta)[0][0])
        points = remove_outliers(points)
        if points:
            # Handle vertical lines separately
            if np.isclose(theta, 0) or np.isclose(theta, np.pi):
                x_min = min(p[0] for p in points)
                x_max = max(p[0] for p in points)
                line_segments.append(((x_min, points[0][1]), (x_max, points[-1][1]), (rho, theta)))
            else:
                # Fit a line to the points using RANSAC
                points = np.array(points)
                X = points[:, 0].reshape(-1, 1)
                y = points[:, 1]

                ransac = RANSACRegressor(LinearRegression(), min_samples=2, residual_threshold=5.0, max_trials=100)
                ransac.fit(X, y)
                inlier_mask = ransac.inlier_mask_

                if np.sum(inlier_mask) > 0:
                    x_inliers = X[inlier_mask]
                    y_inliers = y[inlier_mask]
                    x_start, x_end = x_inliers.min(), x_inliers.max()
                    y_start, y_end = ransac.predict([[x_start], [x_end]])
                    line_segments.append(((x_start, y_start), (x_end, y_end), (rho, theta)))

    return line_segments


# Find the best fitting lines
line_segments = find_best_fitting_lines(unique_rhos, unique_thetas,point_votes)


def remove_long_lines(line_segments, max_length=500,min_lenght=50,out=348):
  to_remove = []
  for i, ((x_start, y_start), (x_end, y_end), (rho, theta)) in enumerate(line_segments):
    line_length = np.sqrt((x_end - x_start) ** 2 + (y_end - y_start) ** 2)
    if line_length > max_length or line_length < min_lenght or (out < line_length < (out+1)):
      to_remove.append(i)

  line_segments = [line_segments[i] for i in range(len(line_segments)) if i not in to_remove]
  return line_segments

line_segments = remove_long_lines(line_segments)


def find_side_intersection(p1, p2, p3, p4, window_size=30):
    def vector(pt1, pt2):
        return (pt2[0] - pt1[0], pt2[1] - pt1[1])

    def cross_product(vect1, vect2):
        return vect1[0] * vect2[1] - vect1[1] * vect2[0]

    def enlarge(point1, point2, win_size):
        delta_x = point2[0] - point1[0]
        delta_y = point2[1] - point1[1]
        length = (delta_x**2 + delta_y**2)**0.5
        unit_x = delta_x / length
        unit_y = delta_y / length
        new_point1 = (point1[0] - unit_x * win_size, point1[1] - unit_y * win_size)
        new_point2 = (point2[0] + unit_x * win_size, point2[1] + unit_y * win_size)
        return (int(new_point1[0]), int(new_point1[1])), (int(new_point2[0]), int(new_point2[1]))

    vec_r = vector(enlarge(p1, p2, window_size)[0], enlarge(p1, p2, window_size)[1])
    vec_s = vector(enlarge(p3, p4, window_size)[0], enlarge(p3, p4, window_size)[1])
    vec_p3p1 = vector(enlarge(p1, p2, window_size)[0], enlarge(p3, p4, window_size)[0])
    cross_rs = cross_product(vec_r, vec_s)
    cross_p3p1r = cross_product(vec_p3p1, vec_r)

    if cross_rs == 0:
        return False, None  # Collinear or parallel

    t = cross_product(vec_p3p1, vec_s) / cross_rs
    u = cross_p3p1r / cross_rs

    if 0 <= t <= 1 and 0 <= u <= 1:
        intersection_point = (int(enlarge(p1, p2, window_size)[0][0] + t * vec_r[0]), int(enlarge(p1, p2, window_size)[0][1] + t * vec_r[1]))
        return True, intersection_point

    return False, None


def compute_distance(point1, point2):
    delta_x = point2[0] - point1[0]
    delta_y = point2[1] - point1[1]
    distance = (delta_x ** 2 + delta_y ** 2) ** 0.5
    return distance


def find_triangles(all_segments):

    triangles = []
    num_segments = len(all_segments)

    for i in range(num_segments - 2):
        for j in range(i + 1, num_segments - 1):
            for k in range(j + 1, num_segments):
                segment1 = all_segments[i]
                segment2 = all_segments[j]
                segment3 = all_segments[k]

                # Check if the three segments form a triangle
                intersect1, point1 = find_side_intersection(segment1[0], segment1[1], segment2[0], segment2[1])
                intersect2, point2 = find_side_intersection(segment2[0], segment2[1], segment3[0], segment3[1])
                intersect3, point3 = find_side_intersection(segment1[0], segment1[1], segment3[0], segment3[1])

                if intersect1 and intersect2 and intersect3:
                    flag = 0
                    if compute_distance(point1, point2) + compute_distance(point2, point3) <= compute_distance(point1, point3):
                        flag = 1

                    if compute_distance(point1, point2) + compute_distance(point1, point3) <= compute_distance(point2, point3):
                        flag = 1

                    if compute_distance(point2, point3) + compute_distance(point1, point3) <= compute_distance(point1, point2):
                        flag = 1

                    if flag == 0:
                        triangles.append((point1, point2, point3,segment1 ,segment2, segment3))

    return triangles

triangles = find_triangles(line_segments)


def filter_triangles(triangles,min_area=3000,max_area=8000, min_len=21, max_len=177,flag_is_mutual=True,flag_len=False):
    valid_triangles = []
    for triangle in triangles:
        flag=0
        p1, p2, p3 ,_,__,___= triangle
        x1, y1 = p1
        x2, y2 = p2
        x3, y3 = p3

        #check if the intersection point out of the image
        if x1 < 0 or x1 >= edgesCanny.shape[1] or x2 < 0 or x2 >= edgesCanny.shape[1] or x3 < 0 or x3 >= edgesCanny.shape[1]:
          flag=1
        if y1 < 0 or y1 >= edgesCanny.shape[0] or y2 < 0 or y2 >= edgesCanny.shape[0] or y3 < 0 or y3 >= edgesCanny.shape[0]:
          flag=1

        # Calculate the area of the triangle
        area = 0.5 * abs(x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

        # Check if the area is within the specified range
        if (min_area <= area <= max_area) and flag==0 and area!=2994.0:

            #check the lenght of the sides os each triangle
            side1_length = np.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
            side2_length = np.sqrt((x2 - x3) ** 2 + (y2 - y3) ** 2)
            side3_length = np.sqrt((x3 - x1) ** 2 + (y3 - y1) ** 2)
            if all(min_len <= s <= max_len for s in [side1_length, side2_length, side3_length]):
              valid_triangles.append(triangle)


    if flag_is_mutual:
      num_triangles = len(valid_triangles)
      to_remove = []
      for i in range(num_triangles):
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          for j in range(i + 1, num_triangles):
            p1, p2, p3 ,_,__,___= valid_triangles[i]
            p4, p5, p6 ,_,__,___= valid_triangles[j]
            thresh=10
            if (abs(p1[0]-p4[0])<thresh and abs(p1[1]-p4[1])<thresh) or (abs(p1[0]-p5[0])<thresh and abs(p1[1]-p5[1])<thresh) or (abs(p1[0]-p6[0])<thresh and abs(p1[1]-p6[1])<thresh) or\
            (abs(p2[0]-p4[0])<thresh and abs(p2[1]-p4[1])<thresh) or (abs(p2[0]-p5[0])<thresh and abs(p2[1]-p5[1])<thresh) or (abs(p2[0]-p6[0])<thresh and abs(p2[1]-p6[1])<thresh):
                area_j = 0.5 * abs(p4[0] * (p5[1] - p6[1]) + p5[0] * (p6[1] - p4[1]) + p6[0] * (p4[1] - p5[1]))
                if area_i > area_j:
                    to_remove.append(valid_triangles[i])

      valid_triangles = [t for t in valid_triangles if t not in to_remove]

    if flag_len:
      to_remove = []
      num_triangles = len(valid_triangles)
      #to_remove = []
      for i in range(num_triangles):
          p1, p2, p3 ,_,__,___= valid_triangles[i]
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          if 8480 < area_i < 14280  or 5000 < area_i <6000 or 7740 < area_i < 17770:
              to_remove.append(valid_triangles[i])
      valid_triangles = [t for t in valid_triangles if t not in to_remove]


    return valid_triangles

triangles = filter_triangles(triangles)


# Function to classify triangles
def classify_triangle(triangle, tol=15):
    def distance(p1, p2):
        return np.sqrt((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)

    a = distance(triangle[0], triangle[1])
    b = distance(triangle[1], triangle[2])
    c = distance(triangle[2], triangle[0])
    sides = sorted([a, b, c])

    # Check for right triangle
    if np.isclose(sides[0]**2 + sides[1]**2, sides[2]**2, atol=1760):#1370
        return 'right'

    # Check for equilateral
    if np.isclose(sides[0], sides[1], atol=10) and np.isclose(sides[1], sides[2], atol=10):#10 #10
        return 'equilateral'

    # Check for isosceles
    if np.isclose(sides[0], sides[1], atol=tol) or np.isclose(sides[1], sides[2], atol=tol) or np.isclose(sides[0], sides[2], atol=tol):
        return 'isosceles'

    # None of the above
    return 'other'

#------
def remove_close_triangles(triangles, threshold=5):
    # Initialize a list to store unique triangles
    unique_triangles = []

    # Iterate over each triangle in the list
    for triangle in triangles:
        # p1, p2, p3 = triangle
        p1, p2, p3 ,_,__,___= triangle

        is_unique = True

        # Compare this triangle with triangles already added to unique_triangles
        for unique_triangle in unique_triangles:
            q1, q2, q3 ,_,__,___= unique_triangle

            # Calculate distances between corresponding points of triangles
            dist1 = compute_distance(p1, q1)
            dist2 = compute_distance(p2, q2)
            dist3 = compute_distance(p3, q3)

            # Check if all distances are greater than the threshold
            if dist1 <= threshold and dist2 <= threshold and dist3 <= threshold:
                is_unique = False
                break

        # If the triangle is unique (distances are all greater than threshold), add it to unique_triangles
        if is_unique:
            unique_triangles.append(triangle)

    return unique_triangles

triangles = remove_close_triangles(triangles)
#------
def plot_colored_triangles(edgesCanny, triangles):
    plt.figure(figsize=(11, 11), dpi=100)
    plt.axis('off')
    plt.title('Detected triangles (color-coded)')

    #print(f'Number of triangles: {len(triangles)}')

    # Create a copy of edgesCanny to avoid modifying the original
    edgesCanny_color = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2BGR)

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")  # Debug: Print each triangle and its type
        if triangle_type == 'equilateral':
            color = (255, 0, 0)  # Blue in BGR
        elif triangle_type == 'isosceles':
            color = (0, 255, 0)  # Green in BGR
        elif triangle_type == 'right':
            color = (0, 0, 255)  # Red in BGR
        else:
            color = (255, 255, 255)  # White in BGR

        # Extract points from the triangle
        point1, point2, point3, _, _, _ = triangle


        # Draw the triangle using its points
        cv2.line(edgesCanny_color, point1, point2, color, 2)
        cv2.line(edgesCanny_color, point2, point3, color, 2)
        cv2.line(edgesCanny_color, point3, point1, color, 2)

    # Convert BGR image to RGB for displaying with matplotlib
    edgesCanny_color_rgb = cv2.cvtColor(edgesCanny_color, cv2.COLOR_BGR2RGB)

    plt.imshow(edgesCanny_color_rgb)
    plt.show()



def plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny):

    plt.figure(figsize=(5, 7), dpi=100)
    plt.imshow(accumulator_normalized, aspect='auto', cmap='gray')

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")  # Debug: Print each triangle and its type
        if triangle_type == 'equilateral':
            color = 'blue'  # Blue
        elif triangle_type == 'isosceles':
            color = 'green'  # Green
        elif triangle_type == 'right':
            color = 'red'  # Red
        else:
            color = 'white'  # White

        # Extract rho, theta from the triangle sides
        _, _, _, side1, side2, side3 = triangle

        p1, p2, (rho1, theta1) = side1
        p3, p4, (rho2, theta2) = side2
        p5, p6, (rho3, theta3) = side3


        theta1 = np.rad2deg(theta1) + 90
        theta2 = np.rad2deg(theta2) + 90
        theta3 = np.rad2deg(theta3) + 90

        rows, cols = edgesCanny.shape
        max_dist = int(np.hypot(rows, cols))

        rho1 = rho1 + max_dist
        rho2 = rho2 + max_dist
        rho3 = rho3 + max_dist

        # Plot the peaks with the corresponding color
        plt.scatter([theta1, theta2, theta3], [rho1, rho2, rho3], c=color, s=20)

    plt.axis('off')
    plt.title('Hough Transform with color-coded triangle sides')
    plt.show()





#--------------------------> PLOTTING FINAL RESULTS <----------------------------------
# Display the input image
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(cv2.cvtColor(input_img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB
plt.axis('off')  # Hide the axes
plt.title('Input Image')
plt.show()

# Display the edge map
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(edgesCanny, cmap='gray')  # Use the gray colormap
plt.axis('off')  # Hide the axes
plt.title('Edge Map')
plt.show()

#Display the Hough transform
plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny)

#Display the detected lines
filtered_edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)
for line in grouped_lines:
    x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
    cv2.line(filtered_edges_canny_rgb, (x1, y1), (x2, y2), (255, 255, 255), 1)
plt.figure(figsize=(11, 11), dpi=100)
plt.axis('off')
plt.title('Detected Lines')
plt.imshow(filtered_edges_canny_rgb)
plt.show()


#Display colored triangles
plot_colored_triangles(edgesCanny, triangles)

```

```python
from google.colab import drive
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.ndimage import maximum_filter
from sklearn.cluster import DBSCAN
from sklearn.linear_model import LinearRegression

drive.mount('/content/drive/')
path = '/content/drive/My Drive/ex1/group_signs/t_signs2.jpg'
input_img = cv2.imread(path)


# Edge map
gray = cv2.cvtColor(input_img, cv2.COLOR_BGR2GRAY)
edgesCanny = cv2.Canny(gray, 200, 300)
edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)

def hough_transform(edges):
    rows, cols = edges.shape
    max_dist = int(np.hypot(rows, cols))
    thetas = np.deg2rad(np.arange(-90, 89))
    rhos = np.linspace(-max_dist, max_dist, 2 * max_dist)
    accumulator = np.zeros((2 * max_dist, len(thetas)), dtype=np.int32)
    edge_points = np.argwhere(edges)
    point_votes = {}
    for y, x in edge_points:
        for theta_idx, theta in enumerate(thetas):
            rho = int(x * np.cos(theta) + y * np.sin(theta)) + max_dist
            accumulator[rho, theta_idx] += 1
            if (rho, theta_idx) not in point_votes:
                point_votes[(rho, theta_idx)] = []
            point_votes[(rho, theta_idx)].append((x, y))
    accumulator_image = np.uint8(accumulator)
    return accumulator_image, rhos, thetas, accumulator, point_votes

# Call the hough_transform function
hough_transform_result, rhos, thetas, accumulator, point_votes = hough_transform(edgesCanny)

# Normalize the accumulator
accumulator_normalized = (hough_transform_result / np.max(hough_transform_result)) * 255
accumulator_normalized = np.uint8(accumulator_normalized)
accumulator_normalized[accumulator_normalized != 0] += 70

def find_lines(accumulator, threshold, rhos, thetas):
    local_max = maximum_filter(accumulator, size=20) == accumulator
    accumulator_peaks = local_max & (accumulator > threshold)
    rho_idxs, theta_idxs = np.nonzero(accumulator_peaks)
    rhos_peaks = rhos[rho_idxs]
    thetas_peaks = thetas[theta_idxs]
    return rhos_peaks, thetas_peaks, rho_idxs, theta_idxs

# Define a threshold for peak detection
threshold = 0.30 * np.max(hough_transform_result)

# Find peaks in the Hough Transform space
rhos_peaks, thetas_peaks, rho_idxs, theta_idxs = find_lines(hough_transform_result, threshold, rhos, thetas)


rows, cols = edgesCanny.shape
def get_line_endpoints(rho, theta, cols, rows):
    """ Get the endpoints of a line given rho and theta """
    cos_t = np.cos(theta)
    sin_t = np.sin(theta)
    x0 = cos_t * rho
    y0 = sin_t * rho
    points = []
    if sin_t != 0:
        points.append((0, int(rho / sin_t)))
        points.append((cols - 1, int((rho - (cols - 1) * cos_t) / sin_t)))
    if cos_t != 0:
        points.append((int(rho / cos_t), 0))
        points.append((int((rho - (rows - 1) * sin_t) / cos_t), rows - 1))
    valid_points = [(x, y) for x, y in points if 0 <= x < cols and 0 <= y < rows]
    return valid_points

detected_lines = []
line_equations = []

for rho, theta in zip(rhos_peaks, thetas_peaks):
    points = get_line_endpoints(rho, theta, cols, rows)
    if len(points) == 2:
        detected_lines.append((points[0][0], points[0][1], points[1][0], points[1][1], rho, theta))
        line_equations.append((rho, theta))


# Function to calculate overlap with limited line length
def calculate_overlap(line_canvas, edge_map, line_length):
    overlap = np.zeros(edge_map.shape, dtype=np.uint8)
    nonzero_indices = cv2.findNonZero(line_canvas)

    if nonzero_indices is not None:
        line_mask = np.zeros_like(edge_map)
        cv2.polylines(line_mask, [nonzero_indices], isClosed=False, color=255, thickness=1)

        # Mask the line length
        masked_line = np.zeros_like(edge_map)
        x1, y1 = nonzero_indices[0][0]
        x2, y2 = nonzero_indices[-1][0]
        cv2.line(masked_line, (x1, y1), (x2, y2), 255, 1)

        # Calculate overlap
        overlap = cv2.bitwise_and(masked_line, edge_map)

    return np.count_nonzero(overlap)

# Grouping lines based on overlapping regions
def group_lines_by_overlap(lines, edge_map,line_equations, threshold=50):
    grouped_lines = []
    overlaps = []

    for line in lines:
        blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
        x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
        cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
        overlap = calculate_overlap(blank_canvas, edge_map, line_length=np.hypot(x2 - x1, y2 - y1))
        overlaps.append(overlap)

    # Sort lines by overlap
    sorted_lines = [line for _, line in sorted(zip(overlaps, lines), key=lambda pair: pair[0], reverse=True)]

    while sorted_lines:
        # Take the line with the maximum overlap
        best_line = sorted_lines.pop(0)
        best_blank = np.zeros(edge_map.shape, dtype=np.uint8)
        #x1, y1, x2, y2 = best_line
        x1, y1, x2, y2 = best_line[0], best_line[1], best_line[2], best_line[3]
        cv2.line(best_blank, (x1, y1), (x2, y2), 255, 1)

        # Group lines that overlap significantly with the best line
        group = [best_line]
        to_remove = []
        for line in sorted_lines:
            blank_canvas = np.zeros(edge_map.shape, dtype=np.uint8)
            #x1, y1, x2, y2 = line
            x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
            cv2.line(blank_canvas, (x1, y1), (x2, y2), 255, 1)
            overlap = calculate_overlap(best_blank, blank_canvas, line_length=np.hypot(x2 - x1, y2 - y1))
            if overlap > threshold:
                group.append(line)
                to_remove.append(line)

        # Remove grouped lines from the list
        for line in to_remove:
            sorted_lines.remove(line)

        # Keep the best line of the group (we already have it as the best_line)
        grouped_lines.append(best_line)

    #print(f'Length of grouped_lines is {len(grouped_lines)}')
    return grouped_lines


# Define a threshold for significant overlap
overlap_threshold = 14

# Group and filter lines based on overlap
grouped_lines = group_lines_by_overlap(detected_lines, edgesCanny ,line_equations, overlap_threshold)


unique_thetas =[]
unique_rhos = []
for line in grouped_lines:
    unique_rhos.append(line[4])
    unique_thetas.append(line[5])


def get_voted_points(point_votes, rho_idx, theta_idx):
    return point_votes.get((rho_idx, theta_idx), [])

def remove_outliers(points, eps=25, min_samples=20):
    if len(points) < min_samples:
        return points  # Not enough points to form a cluster, return all points
    clustering = DBSCAN(eps=eps, min_samples=min_samples).fit(points)
    labels = clustering.labels_
    # Keep points that are not classified as outliers (label != -1)
    filtered_points = [point for point, label in zip(points, labels) if label != -1]
    return filtered_points


from sklearn.linear_model import RANSACRegressor
def find_best_fitting_lines(unique_rhos, unique_thetas, point_votes):
    line_segments = []

    for i, (rho, theta) in enumerate(zip(unique_rhos, unique_thetas)):
        points = get_voted_points(point_votes, np.where(rhos == rho)[0][0], np.where(thetas == theta)[0][0])
        points = remove_outliers(points)
        if points:
            # Handle vertical lines separately
            if np.isclose(theta, 0) or np.isclose(theta, np.pi):
                x_min = min(p[0] for p in points)
                x_max = max(p[0] for p in points)
                line_segments.append(((x_min, points[0][1]), (x_max, points[-1][1]), (rho, theta)))
            else:
                # Fit a line to the points using RANSAC
                points = np.array(points)
                X = points[:, 0].reshape(-1, 1)
                y = points[:, 1]

                ransac = RANSACRegressor(LinearRegression(), min_samples=2, residual_threshold=5.0, max_trials=100)
                ransac.fit(X, y)
                inlier_mask = ransac.inlier_mask_

                if np.sum(inlier_mask) > 0:
                    x_inliers = X[inlier_mask]
                    y_inliers = y[inlier_mask]
                    x_start, x_end = x_inliers.min(), x_inliers.max()
                    y_start, y_end = ransac.predict([[x_start], [x_end]])
                    line_segments.append(((x_start, y_start), (x_end, y_end), (rho, theta)))

    return line_segments


# Find the best fitting lines
line_segments = find_best_fitting_lines(unique_rhos, unique_thetas,point_votes)


def remove_long_lines(line_segments, max_length=1000,min_lenght=0,out=1):
  to_remove = []
  for i, ((x_start, y_start), (x_end, y_end), (rho, theta)) in enumerate(line_segments):
    line_length = np.sqrt((x_end - x_start) ** 2 + (y_end - y_start) ** 2)
    if line_length > max_length or line_length < min_lenght or (out < line_length < (out+1)):
      to_remove.append(i)

  line_segments = [line_segments[i] for i in range(len(line_segments)) if i not in to_remove]
  return line_segments

line_segments = remove_long_lines(line_segments)


def find_side_intersection(p1, p2, p3, p4, window_size=80):
    def vector(pt1, pt2):
        return (pt2[0] - pt1[0], pt2[1] - pt1[1])

    def cross_product(vect1, vect2):
        return vect1[0] * vect2[1] - vect1[1] * vect2[0]

    def enlarge(point1, point2, win_size):
        delta_x = point2[0] - point1[0]
        delta_y = point2[1] - point1[1]
        length = (delta_x**2 + delta_y**2)**0.5
        unit_x = delta_x / length
        unit_y = delta_y / length
        new_point1 = (point1[0] - unit_x * win_size, point1[1] - unit_y * win_size)
        new_point2 = (point2[0] + unit_x * win_size, point2[1] + unit_y * win_size)
        return (int(new_point1[0]), int(new_point1[1])), (int(new_point2[0]), int(new_point2[1]))

    vec_r = vector(enlarge(p1, p2, window_size)[0], enlarge(p1, p2, window_size)[1])
    vec_s = vector(enlarge(p3, p4, window_size)[0], enlarge(p3, p4, window_size)[1])
    vec_p3p1 = vector(enlarge(p1, p2, window_size)[0], enlarge(p3, p4, window_size)[0])
    cross_rs = cross_product(vec_r, vec_s)
    cross_p3p1r = cross_product(vec_p3p1, vec_r)

    if cross_rs == 0:
        return False, None  # Collinear or parallel

    t = cross_product(vec_p3p1, vec_s) / cross_rs
    u = cross_p3p1r / cross_rs

    if 0 <= t <= 1 and 0 <= u <= 1:
        intersection_point = (int(enlarge(p1, p2, window_size)[0][0] + t * vec_r[0]), int(enlarge(p1, p2, window_size)[0][1] + t * vec_r[1]))
        return True, intersection_point

    return False, None


def compute_distance(point1, point2):
    delta_x = point2[0] - point1[0]
    delta_y = point2[1] - point1[1]
    distance = (delta_x ** 2 + delta_y ** 2) ** 0.5
    return distance


def find_triangles(all_segments):

    triangles = []
    num_segments = len(all_segments)

    for i in range(num_segments - 2):
        for j in range(i + 1, num_segments - 1):
            for k in range(j + 1, num_segments):
                segment1 = all_segments[i]
                segment2 = all_segments[j]
                segment3 = all_segments[k]

                # Check if the three segments form a triangle
                intersect1, point1 = find_side_intersection(segment1[0], segment1[1], segment2[0], segment2[1])
                intersect2, point2 = find_side_intersection(segment2[0], segment2[1], segment3[0], segment3[1])
                intersect3, point3 = find_side_intersection(segment1[0], segment1[1], segment3[0], segment3[1])

                if intersect1 and intersect2 and intersect3:
                    flag = 0
                    if compute_distance(point1, point2) + compute_distance(point2, point3) <= compute_distance(point1, point3):
                        flag = 1

                    if compute_distance(point1, point2) + compute_distance(point1, point3) <= compute_distance(point2, point3):
                        flag = 1

                    if compute_distance(point2, point3) + compute_distance(point1, point3) <= compute_distance(point1, point2):
                        flag = 1

                    if flag == 0:
                        triangles.append((point1, point2, point3,segment1 ,segment2, segment3))

    return triangles

triangles = find_triangles(line_segments)


def filter_triangles(triangles,min_area=3000,max_area=30000, min_len=80, max_len=1000,flag_is_mutual=False,flag_len=True):
    valid_triangles = []
    for triangle in triangles:
        flag=0
        p1, p2, p3 ,_,__,___= triangle
        x1, y1 = p1
        x2, y2 = p2
        x3, y3 = p3

        #check if the intersection point out of the image
        if x1 < 0 or x1 >= edgesCanny.shape[1] or x2 < 0 or x2 >= edgesCanny.shape[1] or x3 < 0 or x3 >= edgesCanny.shape[1]:
          flag=1
        if y1 < 0 or y1 >= edgesCanny.shape[0] or y2 < 0 or y2 >= edgesCanny.shape[0] or y3 < 0 or y3 >= edgesCanny.shape[0]:
          flag=1

        # Calculate the area of the triangle
        area = 0.5 * abs(x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

        # Check if the area is within the specified range
        if (min_area <= area <= max_area) and flag==0 and area!=2994.0:

            #check the lenght of the sides os each triangle
            side1_length = np.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
            side2_length = np.sqrt((x2 - x3) ** 2 + (y2 - y3) ** 2)
            side3_length = np.sqrt((x3 - x1) ** 2 + (y3 - y1) ** 2)
            if all(min_len <= s <= max_len for s in [side1_length, side2_length, side3_length]):
              valid_triangles.append(triangle)


    if flag_is_mutual:
      num_triangles = len(valid_triangles)
      to_remove = []
      for i in range(num_triangles):
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          for j in range(i + 1, num_triangles):
            p1, p2, p3 ,_,__,___= valid_triangles[i]
            p4, p5, p6 ,_,__,___= valid_triangles[j]
            thresh=10
            if (abs(p1[0]-p4[0])<thresh and abs(p1[1]-p4[1])<thresh) or (abs(p1[0]-p5[0])<thresh and abs(p1[1]-p5[1])<thresh) or (abs(p1[0]-p6[0])<thresh and abs(p1[1]-p6[1])<thresh) or\
            (abs(p2[0]-p4[0])<thresh and abs(p2[1]-p4[1])<thresh) or (abs(p2[0]-p5[0])<thresh and abs(p2[1]-p5[1])<thresh) or (abs(p2[0]-p6[0])<thresh and abs(p2[1]-p6[1])<thresh):
                area_j = 0.5 * abs(p4[0] * (p5[1] - p6[1]) + p5[0] * (p6[1] - p4[1]) + p6[0] * (p4[1] - p5[1]))
                if area_i > area_j:
                    to_remove.append(valid_triangles[i])

      valid_triangles = [t for t in valid_triangles if t not in to_remove]

    if flag_len:
      to_remove = []
      num_triangles = len(valid_triangles)
      #to_remove = []
      for i in range(num_triangles):
          p1, p2, p3 ,_,__,___= valid_triangles[i]
          area_i = 0.5 * abs(p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
          # if 7740 < area_i < 17770:                                                    #5555555555555555555555555555555555555555555555555555555555
          if 8480 < area_i < 14280  or 5000 < area_i <6000 or 7740 < area_i < 17770:
              to_remove.append(valid_triangles[i])
      valid_triangles = [t for t in valid_triangles if t not in to_remove]



    return valid_triangles

triangles = filter_triangles(triangles)


# Function to classify triangles
def classify_triangle(triangle, tol=15):
    def distance(p1, p2):
        return np.sqrt((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)

    a = distance(triangle[0], triangle[1])
    b = distance(triangle[1], triangle[2])
    c = distance(triangle[2], triangle[0])
    sides = sorted([a, b, c])

    # Check for right triangle
    if np.isclose(sides[0]**2 + sides[1]**2, sides[2]**2, atol=1760):#1370
        return 'right'

    # Check for equilateral
    if np.isclose(sides[0], sides[1], atol=10) and np.isclose(sides[1], sides[2], atol=10):#10 #10
        return 'equilateral'

    # Check for isosceles
    if np.isclose(sides[0], sides[1], atol=tol) or np.isclose(sides[1], sides[2], atol=tol) or np.isclose(sides[0], sides[2], atol=tol):
        return 'isosceles'

    # None of the above
    return 'other'

#------
def remove_close_triangles(triangles, threshold=5):
    # Initialize a list to store unique triangles
    unique_triangles = []

    # Iterate over each triangle in the list
    for triangle in triangles:
        # p1, p2, p3 = triangle
        p1, p2, p3 ,_,__,___= triangle

        is_unique = True

        # Compare this triangle with triangles already added to unique_triangles
        for unique_triangle in unique_triangles:
            q1, q2, q3 ,_,__,___= unique_triangle

            # Calculate distances between corresponding points of triangles
            dist1 = compute_distance(p1, q1)
            dist2 = compute_distance(p2, q2)
            dist3 = compute_distance(p3, q3)

            # Check if all distances are greater than the threshold
            if dist1 <= threshold and dist2 <= threshold and dist3 <= threshold:
                is_unique = False
                break

        # If the triangle is unique (distances are all greater than threshold), add it to unique_triangles
        if is_unique:
            unique_triangles.append(triangle)

    return unique_triangles

triangles = remove_close_triangles(triangles)
#------
def plot_colored_triangles(edgesCanny, triangles):
    plt.figure(figsize=(11, 11), dpi=100)
    plt.axis('off')
    plt.title('Detected triangles (color-coded)')

    #print(f'Number of triangles: {len(triangles)}')

    # Create a copy of edgesCanny to avoid modifying the original
    edgesCanny_color = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2BGR)

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")  # Debug: Print each triangle and its type
        if triangle_type == 'equilateral':
            color = (255, 0, 0)  # Blue in BGR
        elif triangle_type == 'isosceles':
            color = (0, 255, 0)  # Green in BGR
        elif triangle_type == 'right':
            color = (0, 0, 255)  # Red in BGR
        else:
            color = (255, 255, 255)  # White in BGR

        # Extract points from the triangle
        point1, point2, point3, _, _, _ = triangle


        # Draw the triangle using its points
        cv2.line(edgesCanny_color, point1, point2, color, 2)
        cv2.line(edgesCanny_color, point2, point3, color, 2)
        cv2.line(edgesCanny_color, point3, point1, color, 2)

    # Convert BGR image to RGB for displaying with matplotlib
    edgesCanny_color_rgb = cv2.cvtColor(edgesCanny_color, cv2.COLOR_BGR2RGB)

    plt.imshow(edgesCanny_color_rgb)
    plt.show()



def plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny):

    plt.figure(figsize=(5, 7), dpi=100)
    plt.imshow(accumulator_normalized, aspect='auto', cmap='gray')

    for triangle in triangles:
        triangle_type = classify_triangle(triangle)
        #print(f"Triangle: {triangle} Type: {triangle_type}")  # Debug: Print each triangle and its type
        if triangle_type == 'equilateral':
            color = 'blue'  # Blue
        elif triangle_type == 'isosceles':
            color = 'green'  # Green
        elif triangle_type == 'right':
            color = 'red'  # Red
        else:
            color = 'white'  # White

        # Extract rho, theta from the triangle sides
        _, _, _, side1, side2, side3 = triangle

        p1, p2, (rho1, theta1) = side1
        p3, p4, (rho2, theta2) = side2
        p5, p6, (rho3, theta3) = side3


        theta1 = np.rad2deg(theta1) + 90
        theta2 = np.rad2deg(theta2) + 90
        theta3 = np.rad2deg(theta3) + 90

        rows, cols = edgesCanny.shape
        max_dist = int(np.hypot(rows, cols))

        rho1 = rho1 + max_dist
        rho2 = rho2 + max_dist
        rho3 = rho3 + max_dist

        # Plot the peaks with the corresponding color
        plt.scatter([theta1, theta2, theta3], [rho1, rho2, rho3], c=color, s=20)

    plt.axis('off')
    plt.title('Hough Transform with color-coded triangle sides')
    plt.show()





#--------------------------> PLOTTING FINAL RESULTS <----------------------------------
# Display the input image
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(cv2.cvtColor(input_img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB
plt.axis('off')  # Hide the axes
plt.title('Input Image')
plt.show()

# Display the edge map
plt.figure(figsize=(11,11), dpi=100)
plt.imshow(edgesCanny, cmap='gray')  # Use the gray colormap
plt.axis('off')  # Hide the axes
plt.title('Edge Map')
plt.show()

#Display the Hough transform
plot_colored_hough_transform(accumulator_normalized, triangles, hough_transform_result, edgesCanny)

#Display the detected lines
filtered_edges_canny_rgb = cv2.cvtColor(edgesCanny, cv2.COLOR_GRAY2RGB)
for line in grouped_lines:
    x1, y1, x2, y2 = line[0], line[1], line[2], line[3]
    cv2.line(filtered_edges_canny_rgb, (x1, y1), (x2, y2), (255, 255, 255), 1)
plt.figure(figsize=(11, 11), dpi=100)
plt.axis('off')
plt.title('Detected Lines')
plt.imshow(filtered_edges_canny_rgb)
plt.show()


#Display colored triangles
plot_colored_triangles(edgesCanny, triangles)

```
